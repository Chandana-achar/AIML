```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pltb
```

```python
data = [-5,10,15,20,25,30,35,40,45,50,100,200]
```

```python
df = pd.DataFrame(data)
print(df)
```

Outliers detection

```python
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
```

```python
print("Q1:" , Q1)
print("Q3:" , Q3)
print("IQR:" , IQR)
print("lower bonud:" , lower_bound)
print("upper bonud:" , upper_bound)
```

```python
outliers = []
for value in data:
  if value<lower_bound or value>upper_bound:
    outliers.append(value)
print("Outliers:",outliers)
```

```python
plt.boxplot(data)
plt.show()

```
